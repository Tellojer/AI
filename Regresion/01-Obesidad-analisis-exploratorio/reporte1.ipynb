# Resultados
 ## 3.1 Exploración inicial de datos
El conjunto de datos analizado contiene información de 2,111 individuos
y 10 variables relacionadas con características demográficas, hábitos
y condiciones asociadas al nivel de obesidad.

```python
import pandas as pd

data = pd.read_csv("/content/A1.1 Obesidad.csv")
```

La información incluye variables relacionadas con sexo, edad, estatura
y peso, así como indicadores sobre antecedentes familiares de sobrepeso,
hábitos alimenticios, tabaquismo, consumo de vegetales y consumo de
agua. El conjunto incorpora además una variable categórica que
clasifica a cada individuo según su nivel de obesidad.

**Tabla 1. Categorización de datos**

| Variable               | Tipo de dato | Naturaleza   | Descripción                              |
| ---------------------- | ------------ | ------------ | ---------------------------------------- |
| `Sexo`                 | Categórico   | Cualitativa  | Sexo del individuo                       |
| `Edad`                 | Numérico     | Cuantitativa | Edad en años                             |
| `Estatura`             | Numérico     | Cuantitativa | Estatura en metros                       |
| `Peso`                 | Numérico     | Cuantitativa | Peso corporal en kg                      |
| `FamiliarConSobrepeso` | Categórico   | Cualitativa  | Antecedentes familiares de sobrepeso     |
| `ComeMuchasCalorias`   | Categórico   | Cualitativa  | Indicador de consumo elevado de calorías |
| `ComeVegetales`        | Numérico     | Cuantitativa | Nivel registrado de consumo de vegetales |
| `Fumador`              | Categórico   | Cualitativa  | Indicador de tabaquismo                  |
| `ConsumoDeAgua`        | Numérico     | Cuantitativa | Nivel registrado de consumo de agua      |
| `NivelDeObesidad`      | Categórico   | Cualitativa  | Clasificación del nivel de obesidad      |




Una vez identificada la naturaleza técnica de cada variable, es fundamental evaluar la integridad del conjunto de datos antes de proceder con los análisis inferenciales. A continuación, se ejecuta la verificación para validar los datos presentes en la base de datos y confirmar la presencia o
ausencia de valores nulos o faltantes en el total de los registros.

```python
resumen_estructura = pd.DataFrame({
    'Variable': data.columns,
    'Tipo de dato': data.dtypes.values.astype(str),
    'Valores faltantes': data.isnull().sum().values,
    'Porcentaje de faltantes (%)': (data.isnull().sum().values / len(data) * 100).round(2)
})

print("Tabla 2. Estructura y calidad de los datos")
display(resumen_estructura)
```

No se identifica ningun valor faltante en la base de datos en ninguna de las 10 variables presentadas, por lo que no se requiere de imputación previa al análisis.


```python
print("Tabla 3. Descripción de datos cuantitativos")

data.describe().round(2)
```

La base de datos contiene una muestra que incluye personas de entre 14 y 61 años, con una edad promedio de aproximadamente 24 años, por lo que predominan los adultos jóvenes. En cuanto a las características físicas, la estatura promedio es de 1.70 m y el peso promedio es de 86.6 kg. El peso presenta bastante variación entre las personas, con una desviación estándar cercana a 26 kg, lo cual tiene sentido, porque los datos incluyen diferentes niveles de obesidad.

En las variables ComeVegetales y ConsumoDeAgua se encontraron algunos valores decimales, aunque ambas deberían manejarse en una escala del 1 al 3. Esto puede indicar que la base de datos combina información real con datos generados para lograr una mejor distribución entre las diferentes categorías.


```python
variables_categoricas = ['Sexo', 'FamiliarConSobrepeso', 'ComeMuchasCalorias', 'Fumador']

for i, col in enumerate(variables_categoricas, start=4):
    conteo = data[col].value_counts()
    porcentaje = (data[col].value_counts(normalize=True) * 100).round(2)

    tabla_cat = pd.DataFrame({
        'Conteo': conteo,
        'Porcentaje (%)': porcentaje
    })

    print(f"=== Tabla {i}. {col} ===")
    display(tabla_cat)
    print()
```

Al examinar la composición cualitativa de la muestra, destaca una distribución sumamente equitativa en la variable Sexo, con una representación del 50.6% de hombres (1,068 individuos) y 49.4% de mujeres (1,043 individuos). Por otro lado, los indicadores de hábitos y factores de riesgo muestran una concentración: el 81.8% de los participantes reporta antecedentes familiares de exceso de peso (1,726 personas) y el 88.4% admite un consumo frecuente de alimentos con alta densidad calórica (1,866 personas). Por su parte, la prevalencia de consumo de tabaco es marginal, alcanzando apenas al 2.1% del total analizado (44 fumadores).

Esta configuración refleja una selección muestral intencionada, orientada a agrupar a individuos con alta exposición a determinantes genéticos y dietéticos vinculados con el desarrollo de obesidad.

## 3.2 Reflexión sobre el diseño del estudio

Si yo formase parte del equipo, propondría, desde una perspectiva analítica sobre los determinantes del equilibrio energético y la composición corporal, s la incorporación de la variable "Número diario de pasos".

* Justificación: Aunque la base de datos contempla aspectos dietéticos y de consumo, no cuenta con una medición continua del gasto calórico derivado de la actividad física cotidiana no estructurada (conocida como *NEAT* o termogénesis por actividad no vinculada al ejercicio). La medición del número de pasos aportaría un indicador preciso del nivel de sedentarismo o movilidad del participante.
Esta variable permitiría diferenciar de manera objetiva si parte del sobrepeso en los individuos es consecuencia directa y prioritaria al exceso calorias o a la falta de actividad física diaria, ofreciendo un control cuantitativo más sólido al momento de modelar e inferir las causas del exceso de peso en la población estudiada.

* Naturaleza de la variable: Se trata de una variable de tipo cuantitativo discreto, correspondiente al recuento o conteo acumulado de pasos efectuados por el individuo durante un lapso de 24 horas.

## 3.2 Definición de una variable de interés para inferencia

Como líder del proyecto, seleccionaría la variable NivelDeObesidad como la variable de salida del estudio.

Justificación: Esta variable sintetiza de manera directa la finalidad del estudio para determinar el estado de la población observada. En un análisis de inferencia estadística, NivelDeObesidad actúa como el resultado de interés que se busca explicar o predecir en función de los factores demográficos, genéticos y comportamentales registrados en los participantes.

Naturaleza de la variable: Es una variable cualitativa ordinal, la cual clasifica a los individuos en siete niveles ponderados de acuerdo con su Índice de Masa Corporal (IMC): Insufficient_Weight, Normal_Weight, Overweight_Level_I, Overweight_Level_II, Obesity_Type_I, Obesity_Type_II y Obesity_Type_III.  





```python
frecuencia_abs = data['NivelDeObesidad'].value_counts()
frecuencia_rel = (data['NivelDeObesidad'].value_counts(normalize=True) * 100).round(2)

resumen_obesidad = pd.DataFrame({
    'Cantidad de personas': frecuencia_abs,
    'Porcentaje (%)': frecuencia_rel
})

print("--- Tabla 8. Distribución de frecuencias para NivelDeObesidad ---")
display(resumen_obesidad)
```

Al examinar la distribución de NivelDeObesidad, se aprecia un notable balance entre las siete categorías que la integran.

La clase con mayor frecuencia es Obesity_Type_I representando un 16.63% de los casos (351 registros), mientras que la categoría con menor frecuencia es Insufficient_Weight con un 12.88% (272 registros). La diferencia entre la clase más representada y la menos representada es de apenas 3.75 puntos porcentuales. Esta homogeneidad confirma una muestra adecuadamente balanceada, lo que resulta altamente propicio para tareas de inferencia y clasificación, ya que evita sesgos hacia una categoría predominante.

```python
import matplotlib.pyplot as plt


orden_categorias = [
    'Insufficient_Weight',
    'Normal_Weight',
    'Overweight_Level_I',
    'Overweight_Level_II',
    'Obesity_Type_I',
    'Obesity_Type_II',
    'Obesity_Type_III'
]


frecuencia_ordenada = data['NivelDeObesidad'].value_counts().reindex(orden_categorias)


colores_matlab = [
    '#D3E0EA',
    '#B0CEE0',
    '#8EAECF',
    '#6C8EBF',
    '#4A6B9D',
    '#2B4C7E',
    '#1B365D'
]


plt.figure(figsize=(7, 7), dpi=150)


plt.pie(
    frecuencia_ordenada.values,
    labels=frecuencia_ordenada.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=colores_matlab,
    pctdistance=0.75,
    textprops={'fontsize': 9, 'color': 'black'},
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.2}
)


# Dejar espacio en la parte inferior
plt.subplots_adjust(bottom=0.20)

# Título debajo de la gráfica
plt.figtext(
    0.5, 0.04,
    'Gráfica 1. Distribución porcentual del nivel de obesidad',
    ha='center',
    fontsize=11,
    fontweight='bold'
)

plt.show()
```

## 3.4 Análisis gráfico enfocado en una subpoblación
A continuación se selecciona la subpoblación masculina. Dentro de esta subpoblación, se explora la relación entre la variable de salida NivelDeObesidad (categórica ordinal) y la variable cuantitativa Peso (en kilogramos).

A continuación, se ejecutan las visualizaciones en Python mediante un diagrama de cajas (boxplot) para evaluar la posición, dispersión y presencia de valores atípicos, y un gráfico de barras para comparar la media del peso a lo largo de cada nivel de obesidad.

```python
import pandas as pd
import matplotlib.pyplot as plt


data = pd.read_csv("/content/A1.1 Obesidad.csv")


sub_masculino = data[data['Sexo'] == 'Male'].copy()


orden_categorias = [
    'Insufficient_Weight',
    'Normal_Weight',
    'Overweight_Level_I',
    'Overweight_Level_II',
    'Obesity_Type_I',
    'Obesity_Type_II',
    'Obesity_Type_III'
]


datos_por_categoria = [
    sub_masculino[sub_masculino['NivelDeObesidad'] == cat]['Peso'].dropna()
    for cat in orden_categorias
]


plt.figure(figsize=(10, 6), dpi=150)
ax = plt.gca()


bp = ax.boxplot(
    datos_por_categoria,
    patch_artist=True,
    widths=0.5,
    boxprops=dict(facecolor='#E6E6E6', color='#000080', linewidth=1.2),
    whiskerprops=dict(color='#000080', linewidth=1.2, linestyle='--'),
    capprops=dict(color='#000080', linewidth=1.2),
    medianprops=dict(color='#A00000', linewidth=1.5),
    flierprops=dict(marker='+', markeredgecolor='red', markersize=7)
)


ax.set_xticklabels(
    orden_categorias,
    rotation=25,
    ha='right',
    fontsize=9
)

ax.set_xlabel(
    'Nivel de Obesidad',
    fontsize=11,
    fontweight='bold'
)

ax.set_ylabel(
    'Peso (kg)',
    fontsize=11,
    fontweight='bold'
)


ax.grid(True, linestyle=':', alpha=0.6, color='gray')
ax.set_axisbelow(True)


# Espacio para el título debajo de la gráfica
plt.subplots_adjust(bottom=0.30)

# Título debajo de la gráfica
plt.figtext(
    0.5, 0.06,
    'Gráfica 2. Distribución del peso por nivel de obesidad en la subpoblación masculina',
    ha='center',
    fontsize=11,
    fontweight='bold'
)

plt.show()
```

El diagrama de caja revela patrones claros y coherentes sobre la distribución de la masa corporal en los hombres a través de las distintas categorías:

Tendencia ascendente: Se observa un incremento monótono y sostenido en la mediana del peso a medida que se avanza en la escala ordinal de obesidad. La mediana parte de aproximadamente 55.5 kg en la categoría Insufficient_Weight y se eleva de forma continua hasta superar los 117.8 kg en Obesity_Type_II (y alcanzando 173 kg en el registro aislado de Obesity_Type_III).  

Dispersión: En la mayoría de los niveles, los rangos son relativamente estrechos y homogéneos, situándose entre 6 kg y 8 kg de desviación estándar. Esto indica una variabilidad moderada dentro de una misma clasificación, reflejando que la estatura del individuo modula en qué punto exacto dentro de la categoría de IMC se ubica determinado peso.

Observaciones atípicas y distribución: Se aprecian muy pocos valores atípicos fuera de las cajas (como un caso inferior cercano a 45 kg en Normal_Weight y observaciones puntuales en Obesity_Type_I y Obesity_Type_II). Por su parte, la categoría Obesity_Type_III cuenta únicamente con un individuo masculino en la muestra (173 kg), lo que evidencia que en esta población los casos severos de obesidad de tipo III se concentran casi en su totalidad en el sexo femenino.
